```python
# Constraint-Based Coreference Resolution

text = """John and Mary went to the park.
He brought a ball.
She wanted to play with it.
The dog chased him excitedly.
Finally, they all went home."""

# Entities
entities = {
    "John": {"gender": "male", "number": "singular", "type": "person"},
    "Mary": {"gender": "female", "number": "singular", "type": "person"},
    "ball": {"gender": "neutral", "number": "singular", "type": "object"},
    "dog": {"gender": "neutral", "number": "singular", "type": "animal"}
}

# Coreference resolution using constraints
coreference = {
    "He": "John",
    "She": "Mary",
    "it": "ball",
    "him": "John",
    "they": "John, Mary, dog",
    "all": "John, Mary, dog"
}

print("ORIGINAL TEXT:")
print(text)

print("\nCOREFERENCE RESOLUTION:")
for mention, antecedent in coreference.items():
    print(mention, "->", antecedent)

print("\nFINAL COREFERENCE CHAINS:")
print("John -> He -> him -> they")
print("Mary -> She -> they")
print("ball -> it")
print("dog -> they")

print("\nREWRITTEN PARAGRAPH:")
print("John and Mary went to the park.")
print("John brought a ball.")
print("Mary wanted to play with the ball.")
print("The dog chased John excitedly.")
print("Finally, John, Mary, and the dog all went home.")
```

```python
# Constraint-Based Dialog Generation

responses = [
    "Since your exam is tomorrow, take a short break and then return with a clear focus on one topic at a time. You can concentrate better this way, and you should feel confident about your preparation.",

    "Your exam is tomorrow, so don't worry if you cannot concentrate immediately; take a short break and start again with a simple study goal. Stay focused and remind yourself that you can do your best.",

    "Because your exam is tomorrow, give yourself a short break and then focus on one important topic at a time. Keep going with confidence, and your concentration will improve."
]

required_words = ["exam", "concentrate", "you"]
keywords = ["focus", "break", "confident"]


def check_constraints(response):
    text = response.lower()

    # Sentence count
    sentences = response.count(".")
    length_ok = 2 <= sentences <= 3

    # Required entities
    entity_ok = all(word in text for word in required_words)

    # At least two keywords
    keyword_count = sum(word in text for word in keywords)
    keyword_ok = keyword_count >= 2

    # Positive words
    positive_words = ["confident", "best", "improve", "can", "good"]
    positive_ok = any(word in text for word in positive_words)

    return length_ok and entity_ok and keyword_ok and positive_ok


print("DIALOG RESPONSES\n")

for i, response in enumerate(responses, 1):
    print("Response", i)
    print(response)

    if check_constraints(response):
        print("Status: Satisfies all constraints")
    else:
        print("Status: Does not satisfy all constraints")

    print()
```

```python
# Word Sense Disambiguation

sentence = "The bank by the river flooded after the storm, but it was saved by quick action."

print("Sentence:")
print(sentence)

# Context words
river_words = ["river", "flooded", "storm", "water"]
financial_words = ["money", "account", "loan", "cash", "finance"]

sentence_lower = sentence.lower()

river_score = 0
financial_score = 0

# Check river context
for word in river_words:
    if word in sentence_lower:
        river_score += 1

# Check financial context
for word in financial_words:
    if word in sentence_lower:
        financial_score += 1

print("\nConstraint Scores:")
print("Riverbank score:", river_score)
print("Financial bank score:", financial_score)

# Resolve word sense
if river_score > financial_score:
    meaning = "riverbank"
else:
    meaning = "financial bank"

print("\nResolved Meaning of 'bank':", meaning)

# Predicate logic representation
print("\nPredicate Logic Representation:")
print("bank(b)")
print("river(r)")
print("location(b, r)")
print("storm(s)")
print("flood(b)")
print("after(flood(b), s)")
print("quick(a)")
print("saved(b, a)")

# Paraphrase
print("\nSimple English Paraphrase:")
print("The riverbank near the river was flooded after the storm,")
print("but quick action saved it.")
```
