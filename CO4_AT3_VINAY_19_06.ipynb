```python
# QUESTION 1: CFG Tree vs Dependency Parsing

print("=" * 60)
print("CFG TREE VS DEPENDENCY PARSING")
print("=" * 60)

sentence = "The boy eats an apple"

print("\nSentence:")
print(sentence)

# ---------------------------------------------------------
# CFG representation
# ---------------------------------------------------------

print("\n1. CFG TREE REPRESENTATION")
print("-" * 40)

print("""
              S
            /   \\
          NP     VP
         / \\    /  \\
       Det  N  V    NP
        |   |  |   /  \\
       The boy eats Det  N
                    |    |
                   an  apple
""")

print("CFG represents the hierarchical structure of the sentence.")
print("It shows how words form phrases such as NP and VP.")


# ---------------------------------------------------------
# Dependency representation
# ---------------------------------------------------------

print("\n2. DEPENDENCY REPRESENTATION")
print("-" * 40)

dependencies = [
    ("eats", "root", "ROOT"),
    ("boy", "nsubj", "eats"),
    ("The", "det", "boy"),
    ("apple", "obj", "eats"),
    ("an", "det", "apple")
]

print("Word       Relation       Head")
print("-" * 40)

for word, relation, head in dependencies:
    print(f"{word:<10} {relation:<14} {head}")


# ---------------------------------------------------------
# Comparison
# ---------------------------------------------------------

print("\n3. COMPARISON")
print("-" * 40)

comparison = {
    "CFG": "Represents hierarchical phrase structure",
    "Dependency": "Represents direct relationships between words",
    "CFG": "Useful for grammar and phrase structure",
    "Dependency": "Useful for word-to-word relationships"
}

print("CFG:")
print("- Shows sentence hierarchy")
print("- Groups words into phrases")
print("- Useful for syntactic analysis")

print("\nDependency Parsing:")
print("- Shows head-dependent relationships")
print("- Directly connects related words")
print("- Useful for information extraction")

print("\n4. CONCLUSION")
print("-" * 40)

print("Dependency parsing is more effective when the goal is")
print("to capture direct relationships between words.")
print("CFG is better for understanding phrase hierarchy.")
```

```python
# QUESTION 2: Top-Down vs Earley Parsing

print("=" * 60)
print("TOP-DOWN PARSING VS EARLEY PARSING")
print("=" * 60)

sentence = "Book a flight to Delhi"

print("\nInput:")
print(sentence)


# ---------------------------------------------------------
# Top-Down Parsing
# ---------------------------------------------------------

print("\n1. TOP-DOWN PARSING")
print("-" * 40)

rules = [
    "S -> VP",
    "VP -> V NP",
    "VP -> V NP PP",
    "NP -> DET N",
    "PP -> PREP NP"
]

print("Starting from S:")

for rule in rules:
    print("Trying:", rule)

print("\nTop-down parser starts with the start symbol")
print("and predicts grammar rules.")

print("\nProblems:")
print("- May require backtracking")
print("- Ambiguous sentences can be expensive")
print("- Incomplete input can cause difficulty")
print("- Not ideal for highly dynamic input")


# ---------------------------------------------------------
# Earley Parsing
# ---------------------------------------------------------

print("\n2. EARLEY PARSING")
print("-" * 40)

print("Earley parsing maintains partial parsing states.")

partial_inputs = [
    "Book",
    "Book a flight",
    "Book a flight to",
    "Book a flight to Delhi"
]

for text in partial_inputs:
    print("Input received:", text)
    print("Parser keeps the current partial state.")

print("\nEarley parsing supports:")
print("- Incomplete input")
print("- Ambiguous input")
print("- General CFGs")
print("- Incremental parsing")


# ---------------------------------------------------------
# Comparison
# ---------------------------------------------------------

print("\n3. COMPARISON")
print("-" * 40)

print(f"{'Feature':<25}{'Top-Down':<20}{'Earley'}")
print("-" * 65)

data = [
    ("Ambiguous input", "Difficult", "Good"),
    ("Incomplete input", "Difficult", "Good"),
    ("Backtracking", "High", "Lower"),
    ("General CFG", "Limited", "Good"),
    ("Real-time use", "Less suitable", "More suitable")
]

for feature, topdown, earley in data:
    print(f"{feature:<25}{topdown:<20}{earley}")


# ---------------------------------------------------------
# Conclusion
# ---------------------------------------------------------

print("\n4. CONCLUSION")
print("-" * 40)

print("Earley parsing is more suitable for dynamic applications")
print("because it can handle incomplete and ambiguous input")
print("while maintaining partial parsing information.")
```

```python
# QUESTION 3: CFG vs PCFG vs Neural Parsing

print("=" * 60)
print("CFG VS PCFG VS NEURAL PARSING")
print("=" * 60)

sentence = "She saw the man with a telescope"

print("\nSentence:")
print(sentence)


# ---------------------------------------------------------
# Ambiguity
# ---------------------------------------------------------

print("\n1. AMBIGUITY")
print("-" * 40)

print("Interpretation 1:")
print("She used a telescope to see the man.")

print("\nInterpretation 2:")
print("The man had a telescope.")


# ---------------------------------------------------------
# CFG
# ---------------------------------------------------------

print("\n2. CFG")
print("-" * 40)

print("CFG can generate both possible structures.")

print("""
Structure 1:
She -> saw -> man
             |
             with telescope

Structure 2:
She -> saw -> man
       |
       with telescope
""")

print("CFG identifies possible grammatical structures")
print("but does not know which interpretation is more likely.")


# ---------------------------------------------------------
# PCFG
# ---------------------------------------------------------

print("\n3. PCFG")
print("-" * 40)

parses = {
    "She used the telescope": 0.70,
    "The man had the telescope": 0.30
}

for interpretation, probability in parses.items():
    print(interpretation, "->", probability)

best = max(parses, key=parses.get)

print("\nPCFG selects:")
print(best)


# ---------------------------------------------------------
# Neural Parsing
# ---------------------------------------------------------

print("\n4. NEURAL PARSING")
print("-" * 40)

print("Neural parsers use machine learning models")
print("to understand context and sentence relationships.")

print("\nFor example, the model can consider:")
print("- Word meaning")
print("- Context")
print("- Surrounding words")
print("- Training data")
print("- Sentence relationships")

print("\nA neural parser may identify the most likely")
print("meaning based on the context of the complete sentence.")


# ---------------------------------------------------------
# Comparison
# ---------------------------------------------------------

print("\n5. COMPARISON")
print("-" * 40)

print(f"{'Method':<20}{'Ambiguity Handling'}")
print("-" * 45)
print(f"{'CFG':<20}{'Generates alternatives'}")
print(f"{'PCFG':<20}{'Ranks alternatives'}")
print(f"{'Neural Parsing':<20}{'Uses learned context'}")


# ---------------------------------------------------------
# Conclusion
# ---------------------------------------------------------

print("\n6. CONCLUSION")
print("-" * 40)

print("CFG is useful for basic grammatical structure.")
print("PCFG improves CFG by assigning probabilities.")
print("Neural parsing can use context and learned patterns.")
print("For real-world applications, neural parsing is generally")
print("more powerful, while PCFG is simpler and easier to interpret.")
```

```python
# QUESTION 4: Feature Structures vs Subcategorization Frames

print("=" * 60)
print("FEATURE STRUCTURES VS SUBCATEGORIZATION FRAMES")
print("=" * 60)


# ---------------------------------------------------------
# Feature Structures
# ---------------------------------------------------------

print("\n1. FEATURE STRUCTURES")
print("-" * 40)

subject = {
    "word": "students",
    "person": 3,
    "number": "plural"
}

verb = {
    "word": "study",
    "person": 3,
    "number": "plural"
}

print("Subject:", subject)
print("Verb:", verb)

if (subject["person"] == verb["person"] and
    subject["number"] == verb["number"]):

    print("\nAgreement: CORRECT")

else:

    print("\nAgreement: INCORRECT")


# Test incorrect sentence

print("\nTesting incorrect sentence:")

wrong_verb = {
    "word": "studies",
    "person": 3,
    "number": "singular"
}

if (subject["person"] == wrong_verb["person"] and
    subject["number"] == wrong_verb["number"]):

    print("Agreement: CORRECT")

else:

    print("Agreement: INCORRECT")


# ---------------------------------------------------------
# Subcategorization
# ---------------------------------------------------------

print("\n2. SUBCATEGORIZATION FRAMES")
print("-" * 40)

verb_frames = {
    "eat": ["Subject", "Object"],
    "give": ["Subject", "Object", "Indirect Object"],
    "sleep": ["Subject"],
    "put": ["Subject", "Object", "Location"]
}

for verb, arguments in verb_frames.items():

    print("\nVerb:", verb)
    print("Required arguments:", arguments)


# ---------------------------------------------------------
# Test sentence
# ---------------------------------------------------------

print("\n3. TESTING VERB ARGUMENTS")
print("-" * 40)

verb = "put"

sentence_arguments = [
    "John",
    "the book",
    "on the table"
]

required_arguments = verb_frames[verb]

print("Sentence:")
print("John put the book on the table")

print("\nVerb:", verb)
print("Required:", required_arguments)
print("Provided:", sentence_arguments)

if len(sentence_arguments) >= len(required_arguments):

    print("\nArgument structure: CORRECT")

else:

    print("\nArgument structure: INCOMPLETE")


# ---------------------------------------------------------
# Comparison
# ---------------------------------------------------------

print("\n4. COMPARISON")
print("-" * 40)

print("Feature Structures:")
print("- Handle number")
print("- Handle person")
print("- Handle tense")
print("- Enforce subject-verb agreement")

print("\nSubcategorization Frames:")
print("- Identify verb requirements")
print("- Identify required arguments")
print("- Identify objects and complements")
print("- Describe verb argument structure")


# ---------------------------------------------------------
# Conclusion
# ---------------------------------------------------------

print("\n5. CONCLUSION")
print("-" * 40)

print("Feature Structures are better for enforcing")
print("grammatical agreement.")

print("\nSubcategorization frames are better for")
print("representing verb argument structures.")

print("\nFor a complete grammar system, both should be used together.")
```

```python
# QUESTION 5: Transition-Based vs Graph-Based Parsing

print("=" * 60)
print("TRANSITION-BASED VS GRAPH-BASED PARSING")
print("=" * 60)

sentence = ["The", "boy", "eats", "an", "apple"]

print("\nSentence:")
print(" ".join(sentence))


# ---------------------------------------------------------
# Transition-Based Parsing
# ---------------------------------------------------------

print("\n1. TRANSITION-BASED PARSING")
print("-" * 40)

stack = ["ROOT"]
buffer = sentence.copy()
dependencies = []

print("Initial Stack:", stack)
print("Initial Buffer:", buffer)

# Shift operations

while buffer:

    word = buffer.pop(0)
    stack.append(word)

    print("\nSHIFT:", word)
    print("Stack:", stack)
    print("Buffer:", buffer)

# Create simple dependencies

dependencies = [
    ("eats", "ROOT"),
    ("boy", "eats"),
    ("The", "boy"),
    ("apple", "eats"),
    ("an", "apple")
]

print("\nTransition-Based Dependencies:")

for word, head in dependencies:
    print(word, "->", head)


# ---------------------------------------------------------
# Graph-Based Parsing
# ---------------------------------------------------------

print("\n2. GRAPH-BASED PARSING")
print("-" * 40)

print("Graph-based parsing considers possible")
print("relationships between words.")

possible_edges = [
    ("eats", "ROOT"),
    ("boy", "eats"),
    ("The", "boy"),
    ("apple", "eats"),
    ("an", "apple")
]

print("\nPossible Dependency Edges:")

for dependent, head in possible_edges:
    print(dependent, "->", head)

print("\nThe parser selects the best combination")
print("of dependency edges to form a complete tree.")


# ---------------------------------------------------------
# Decision Making
# ---------------------------------------------------------

print("\n3. DECISION MAKING")
print("-" * 40)

print("Transition-Based:")
print("- Makes local decisions step-by-step")
print("- Uses operations such as SHIFT and REDUCE")
print("- Usually very fast")

print("\nGraph-Based:")
print("- Considers many possible dependency edges")
print("- Finds the best overall dependency tree")
print("- Usually gives stronger global decisions")


# ---------------------------------------------------------
# Performance Comparison
# ---------------------------------------------------------

print("\n4. PERFORMANCE COMPARISON")
print("-" * 40)

print(f"{'Feature':<25}{'Transition':<20}{'Graph-Based'}")
print("-" * 65)

data = [
    ("Decision making", "Local", "Global"),
    ("Speed", "Very fast", "Slower"),
    ("Memory", "Lower", "Higher"),
    ("Large datasets", "Very suitable", "More expensive"),
    ("Global accuracy", "Lower", "Higher"),
    ("Real-time use", "Excellent", "Moderate")
]

for feature, transition, graph in data:
    print(f"{feature:<25}{transition:<20}{graph}")


# ---------------------------------------------------------
# Conclusion
# ---------------------------------------------------------

print("\n5. CONCLUSION")
print("-" * 40)

print("Transition-based parsing is generally more suitable")
print("for large-scale and real-time applications because")
print("it makes fast local decisions and requires less computation.")

print("\nGraph-based parsing can provide better global decisions")
print("but usually requires more computation.")
```
