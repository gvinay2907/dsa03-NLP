```python
#TASK 1

words = ["analyzing", "analysis", "analytical"]

def process(word):
    if word.endswith("ing"):
        root = word[:-3]
        suffix = "-ing"
        form = "Inflectional"
    elif word.endswith("sis"):
        root = word[:-3]
        suffix = "-sis"
        form = "Derivational"
    elif word.endswith("ical"):
        root = word[:-4]
        suffix = "-ical"
        form = "Derivational"
    else:
        root = word
        suffix = "None"
        form = "Base"

    normalized = "analyze"
    return [word, root, suffix, form, normalized]

print("Word\t\tRoot\t\tSuffix\tType\t\tNormalized")
for w in words:
    r = process(w)
    print(f"{r[0]}\t{r[1]}\t\t{r[2]}\t{r[3]}\t{r[4]}")
```

```python
#TASK 2

words = ["disagree", "agreement", "agreeable"]

def parse(word):
    prefix = "None"
    suffix = "None"
    root = word
    form = "Base"

    if word.startswith("dis"):
        prefix = "dis-"
        root = word[3:]
        form = "Derivational (Prefix)"
    elif word.endswith("ment"):
        suffix = "-ment"
        root = word[:-4]
        form = "Derivational (Noun)"
    elif word.endswith("able"):
        suffix = "-able"
        root = word[:-4]
        form = "Derivational (Adjective)"

    normalized = "agree"
    return [word, prefix, root, suffix, form, normalized]

print("Word\t\tPrefix\tRoot\tSuffix\tType\t\t\tNormalized")
for w in words:
    r = parse(w)
    print(f"{r[0]}\t{r[1]}\t{r[2]}\t{r[3]}\t{r[4]}\t{r[5]}")
```

```python
#TASK 3

words = ["govern", "government", "governance"]

def normalize(word):
    root = "govern"
    suffix = "None"
    level = "Level 0"

    if word.endswith("ment"):
        suffix = "-ment"
        level = "Level 1"
    elif word.endswith("ance"):
        suffix = "-ance"
        level = "Level 1"

    normalized = root
    return [word, root, suffix, level, normalized]

print("Word\t\tRoot\tSuffix\tHierarchy\tNormalized")
for w in words:
    r = normalize(w)
    print(f"{r[0]}\t{r[1]}\t{r[2]}\t{r[3]}\t{r[4]}")
```

```python
#TASK 4
words = ["activate", "activation", "reactivation"]

def parse(word):
    prefix = "None"
    suffix = "None"
    root = "activate"
    sequence = "Base Verb"

    if word == "activation":
        suffix = "-ion"
        sequence = "activate + ion"
    elif word == "reactivation":
        prefix = "re-"
        suffix = "-ion"
        sequence = "re + activate + ion"

    normalized = "activate"
    return [word, prefix, root, suffix, sequence, normalized]

print("Word\t\tPrefix\tRoot\tSuffix\tSequence\t\t\tNormalized")
for w in words:
    r = parse(w)
    print(f"{r[0]}\t{r[1]}\t{r[2]}\t{r[3]}\t{r[4]}\t{r[5]}")
```

```python
#TASK 5

words = ["create", "creates", "creating"]

def process(word):
    suffix = "None"
    category = "Base Form"
    root = word

    if word.endswith("s") and word != "create":
        root = word[:-1]
        suffix = "-s"
        category = "Third-person Singular"
    elif word.endswith("ing"):
        root = word[:-3]
        suffix = "-ing"
        category = "Present Participle"

    normalized = "create"
    return [word, suffix, category, root, normalized]

print("Word\t\tSuffix\tCategory\t\t\tRoot\tNormalized")
for w in words:
    r = process(w)
    print(f"{r[0]}\t{r[1]}\t{r[2]}\t{r[3]}\t{r[4]}")
```
