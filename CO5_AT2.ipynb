```python
# Reference Resolution

text = "Ravi met Arun at the library. He borrowed a book and later returned it."

print("Original Text:")
print(text)

# Entities
entities = ["Ravi", "Arun", "book"]

# Pronoun resolution
# Using first-mentioned person as the antecedent for "He"
he = "Ravi"
it = "book"

print("\nResolved Pronouns:")
print("He ->", he)
print("it ->", it)

# Resolved discourse
resolved_text = (
    "Ravi met Arun at the library. "
    "Ravi borrowed a book and later returned the book."
)

print("\nResolved Discourse:")
print(resolved_text)

print("\nValidation:")
print("'He' is resolved as Ravi using discourse context.")
print("'it' is resolved as book because a book can be borrowed and returned.")
```

```python
# Text Coherence and Discourse Structure

text = (
    "The roads were flooded after heavy rainfall. "
    "Therefore, schools were closed for the day. "
    "Students attended classes online."
)

sentences = [
    "The roads were flooded after heavy rainfall.",
    "Therefore, schools were closed for the day.",
    "Students attended classes online."
]

print("Original Text:\n")
print(text)

print("\nDiscourse Relations:")

# Relation 1
if "after" in sentences[0].lower():
    print("Sentence 1: Cause-Effect")
    print("Heavy rainfall -> Roads flooded")

# Relation 2
if "therefore" in sentences[1].lower():
    print("Sentence 1 -> Sentence 2: Cause-Effect")
    print("Roads flooded -> Schools closed")

# Relation 3
print("Sentence 2 -> Sentence 3: Consequence/Sequence")
print("Schools closed -> Students attended online classes")

print("\nDiscourse Structure:")
print("Heavy rainfall")
print("      |")
print("      v")
print("Roads flooded")
print("      |")
print("      | Therefore")
print("      v")
print("Schools closed")
print("      |")
print("      v")
print("Students attended online classes")

print("\nCoherence:")
print("Each sentence logically follows from the previous sentence.")
print("Therefore, the text forms a coherent cause-effect sequence.")
```

```python
# Dialogue Act Identification

conversation = [
    ("User", "Can you book a train ticket for me?"),
    ("Agent", "Sure, where would you like to travel?"),
    ("User", "I want to go to Chennai."),
    ("Agent", "Your ticket has been booked.")
]


def identify_dialogue_act(sentence):
    sentence = sentence.lower()

    if "can you" in sentence and "book" in sentence:
        return "Request"

    elif "where" in sentence or "what" in sentence or "when" in sentence:
        return "Question"

    elif "i want" in sentence or "i need" in sentence:
        return "Inform"

    elif "has been booked" in sentence or "confirmed" in sentence:
        return "Confirmation"

    else:
        return "Action"


print("Dialogue Act Classification:\n")

acts = []

for speaker, sentence in conversation:

    act = identify_dialogue_act(sentence)

    acts.append(act)

    print(speaker + ": " + sentence)
    print("Dialogue Act:", act)
    print()


print("Dialogue Act Sequence:")
print(" -> ".join(acts))

print("\nEvaluation:")
print("Request -> Agent understands that ticket booking is required.")
print("Question -> Agent asks for missing information.")
print("Inform -> User provides the destination.")
print("Confirmation -> Agent confirms the booking.")
```

```python
# Surface Realization

semantic = {
    "action": "buy",
    "agent": "student",
    "object": "book",
    "tense": "past"
}

print("Semantic Representation:")
print(semantic)

# Lexical selection
agent = "The " + semantic["agent"]

# Tense conversion
if semantic["action"] == "buy" and semantic["tense"] == "past":
    verb = "bought"
else:
    verb = semantic["action"]

object_word = "a " + semantic["object"]

# Sentence structure
sentence = agent + " " + verb + " " + object_word + "."

print("\nGenerated Sentence:")
print(sentence)

# Validation
print("\nGrammar Validation:")
print("Subject: The student")
print("Verb: bought")
print("Object: a book")
print("Tense: Past")
print("Structure: Subject + Verb + Object")
print("Result: Grammatically Correct")
```

```python
# Hybrid Interlingua + Statistical Translation

source = "The boy is playing football."

print("Source Sentence:")
print(source)

# -----------------------------------
# Step 1: Source Analysis
# -----------------------------------

print("\n1. Source Analysis")

agent = "boy"
action = "play"
object_word = "football"
tense = "present"
aspect = "progressive"

print("Agent:", agent)
print("Action:", action)
print("Object:", object_word)
print("Tense:", tense)
print("Aspect:", aspect)


# -----------------------------------
# Step 2: Interlingua
# -----------------------------------

print("\n2. Interlingua Representation")

interlingua = (
    "PLAY("
    "Agent=BOY, "
    "Object=FOOTBALL, "
    "Tense=PRESENT, "
    "Aspect=PROGRESSIVE)"
)

print(interlingua)


# -----------------------------------
# Step 3: Candidate Translations
# -----------------------------------

print("\n3. Candidate Translations")

candidates = [
    ("சிறுவன் கால்பந்து விளையாடுகிறான்.", 1.87),
    ("பையன் கால்பந்து விளையாடுகிறான்.", 1.78),
    ("சிறுவன் கால்பந்து விளையாடினான்.", 1.50)
]

for sentence, score in candidates:
    print(sentence, "Score:", score)


# -----------------------------------
# Step 4: Statistical Scoring
# -----------------------------------

print("\n4. Statistical Scoring")

best_sentence = candidates[0][0]
best_score = candidates[0][1]

for sentence, score in candidates:

    if score > best_score:
        best_sentence = sentence
        best_score = score

print("Highest Score:", best_score)


# -----------------------------------
# Step 5: Final Translation
# -----------------------------------

print("\n5. Final Translation")
print(best_sentence)


# -----------------------------------
# Evaluation
# -----------------------------------

print("\nEvaluation:")
print("Interlingua preserves the meaning of the source sentence.")
print("Statistical scoring selects the most suitable translation.")
print("The final translation preserves subject, action, object and tense.")
```
