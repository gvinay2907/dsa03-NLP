```python
# QUESTION 1: Banking Chatbot using CFG, PCFG and Feature Structures

print("=" * 60)
print("BANKING CUSTOMER SUPPORT CHATBOT")
print("=" * 60)

# ---------------------------------------------------------
# 1. Input Query
# ---------------------------------------------------------

query = "Show me the transactions with the card from last month"

print("\nUser Query:")
print(query)


# ---------------------------------------------------------
# 2. CFG Analysis
# ---------------------------------------------------------

print("\n1. CFG ANALYSIS")
print("-" * 40)

print("Grammar:")
print("S  -> VP")
print("VP -> V NP")
print("NP -> NP PP")
print("NP -> DET N")
print("PP -> P NP")

print("\nPossible ambiguity:")
print("The phrase 'with the card' can mean:")
print("1. Transactions made using the card")
print("2. A card associated with the transactions")

print("\nThe phrase 'from last month' can mean:")
print("1. Transactions from last month")
print("2. A card from last month")


# ---------------------------------------------------------
# 3. PCFG Analysis
# ---------------------------------------------------------

print("\n2. PCFG ANALYSIS")
print("-" * 40)

parses = {
    "Transactions made using the card from last month": 0.85,
    "Card associated with transactions from last month": 0.15
}

for meaning, probability in parses.items():
    print(meaning)
    print("Probability:", probability)

best_parse = max(parses, key=parses.get)

print("\nMost Probable Interpretation:")
print(best_parse)


# ---------------------------------------------------------
# 4. Feature Structure
# ---------------------------------------------------------

print("\n3. FEATURE STRUCTURE")
print("-" * 40)

subject = {
    "word": "transactions",
    "number": "plural",
    "person": 3
}

verb = {
    "word": "are",
    "number": "plural",
    "person": 3
}

print("Subject:", subject)
print("Verb:", verb)

if (subject["number"] == verb["number"] and
    subject["person"] == verb["person"]):
    print("Subject-Verb Agreement: Correct")
else:
    print("Subject-Verb Agreement: Incorrect")


# ---------------------------------------------------------
# 5. Earley Parsing Concept
# ---------------------------------------------------------

print("\n4. EARLEY PARSING")
print("-" * 40)

print("Earley parsing is useful for:")
print("- Ambiguous sentences")
print("- Long queries")
print("- Incomplete queries")
print("- Reducing unnecessary backtracking")

print("\nExample partial query:")
print("Show me the transactions with...")

print("Earley parser can maintain the partial structure")
print("and continue when more words are received.")


# ---------------------------------------------------------
# 6. Banking Intent Extraction
# ---------------------------------------------------------

print("\n5. FINAL BANKING INTERPRETATION")
print("-" * 40)

result = {
    "intent": "show_transactions",
    "payment_method": "card",
    "time_period": "last_month"
}

for key, value in result.items():
    print(key, ":", value)


# ---------------------------------------------------------
# 7. Final Conclusion
# ---------------------------------------------------------

print("\n6. CONCLUSION")
print("-" * 40)

print("CFG provides grammatical structure.")
print("PCFG resolves ambiguity using probabilities.")
print("Feature Structures check subject-verb agreement.")
print("Earley Parsing handles complex and partial queries.")
print("Together, these techniques improve banking chatbot accuracy.")
```

```python
# QUESTION 2: Voice Assistant using Top-Down and Earley Parsing Concepts

print("=" * 60)
print("VOICE ASSISTANT - FLIGHT BOOKING")
print("=" * 60)

# ---------------------------------------------------------
# 1. Input Command
# ---------------------------------------------------------

command = "Book a flight to Delhi with a window seat"

print("\nVoice Command:")
print(command)


# ---------------------------------------------------------
# 2. Ambiguity Analysis
# ---------------------------------------------------------

print("\n1. AMBIGUITY ANALYSIS")
print("-" * 40)

print("Possible Interpretation 1:")
print("Book a flight to Delhi")
print("Seat preference = Window")

print("\nPossible Interpretation 2:")
print("Book a flight to Delhi that has window seats available")


# ---------------------------------------------------------
# 3. Top-Down Parsing
# ---------------------------------------------------------

print("\n2. TOP-DOWN PARSING")
print("-" * 40)

rules = [
    "S -> NP VP",
    "S -> VP",
    "VP -> V NP",
    "VP -> V NP PP"
]

print("Parser starts from the start symbol S.")

for rule in rules:
    print("Trying:", rule)

print("\nIf the selected rule fails,")
print("the parser may need to backtrack.")

print("\nProblems:")
print("- Backtracking")
print("- Slow for ambiguous input")
print("- Difficult with incomplete commands")
print("- Not ideal for real-time voice systems")


# ---------------------------------------------------------
# 4. Earley Parsing Concept
# ---------------------------------------------------------

print("\n3. EARLEY PARSING")
print("-" * 40)

print("Earley parsing maintains partial parsing states.")

partial_command = "Book a flight to Delhi with..."

print("Partial Input:")
print(partial_command)

print("\nEarley parser keeps the current parsing state")
print("and continues when the user provides more input.")

print("\nExample:")
print("New input: 'a window seat'")
print("Parser continues from the previous state.")


# ---------------------------------------------------------
# 5. PCFG Ranking
# ---------------------------------------------------------

print("\n4. PCFG RANKING")
print("-" * 40)

interpretations = {
    "Book flight to Delhi with window seat": 0.90,
    "Book flight to Delhi with window seat availability": 0.10
}

for meaning, probability in interpretations.items():
    print(meaning)
    print("Probability:", probability)

best = max(interpretations, key=interpretations.get)

print("\nSelected Interpretation:")
print(best)


# ---------------------------------------------------------
# 6. Intent and Entity Extraction
# ---------------------------------------------------------

print("\n5. INTENT AND ENTITY EXTRACTION")
print("-" * 40)

result = {
    "intent": "book_flight",
    "destination": "Delhi",
    "seat_preference": "Window"
}

for key, value in result.items():
    print(key, ":", value)


# ---------------------------------------------------------
# 7. Comparison
# ---------------------------------------------------------

print("\n6. TOP-DOWN VS EARLEY")
print("-" * 40)

comparison = {
    "Ambiguity": ["Poor", "Good"],
    "Backtracking": ["High", "Lower"],
    "Partial Input": ["Difficult", "Good"],
    "Real-Time Use": ["Limited", "Better"]
}

print("Feature              Top-Down       Earley")
print("-" * 50)

for feature, values in comparison.items():
    print(f"{feature:<20} {values[0]:<15} {values[1]}")


# ---------------------------------------------------------
# 8. Final Conclusion
# ---------------------------------------------------------

print("\n7. CONCLUSION")
print("-" * 40)

print("Top-down parsing may require repeated backtracking.")
print("Earley parsing handles ambiguity and partial input better.")
print("Therefore, Earley parsing is more suitable for")
print("real-time voice assistant applications.")
```

```python
# QUESTION 3: Healthcare NLP System
# CFG + PCFG + Feature Structures + Earley Concept
# + Sub-categorization + Information Extraction

import re
import json

print("=" * 60)
print("HEALTHCARE NLP SYSTEM")
print("=" * 60)


# ---------------------------------------------------------
# 1. Input Medical Sentence
# ---------------------------------------------------------

text = """
The doctor who reviewed the patient last week recommends
starting medication and scheduling a follow-up visit in Chennai.
"""

text = " ".join(text.split())

print("\n1. INPUT")
print("-" * 40)
print(text)


# ---------------------------------------------------------
# 2. Tokenization
# ---------------------------------------------------------

print("\n2. TOKENIZATION")
print("-" * 40)

tokens = text.replace(".", "").split()

print(tokens)


# ---------------------------------------------------------
# 3. Medical Entity Extraction
# ---------------------------------------------------------

print("\n3. MEDICAL ENTITY EXTRACTION")
print("-" * 40)

entities = {
    "doctor": "Medical Person",
    "patient": "Patient",
    "medication": "Treatment",
    "last week": "Time",
    "Chennai": "Location"
}

for entity, entity_type in entities.items():
    print(entity, "->", entity_type)


# ---------------------------------------------------------
# 4. CFG Analysis
# ---------------------------------------------------------

print("\n4. CFG ANALYSIS")
print("-" * 40)

print("Basic Grammar:")
print("S  -> NP VP")
print("NP -> DET N")
print("NP -> DET N RC")
print("RC -> PRON VP")
print("VP -> V NP")
print("VP -> V VP")
print("VP -> V NP CONJ VP")
print("PP -> PREP NP")

print("\nSentence Structure:")
print("Subject: The doctor")
print("Relative Clause: who reviewed the patient last week")
print("Main Verb: recommends")
print("Action 1: starting medication")
print("Action 2: scheduling a follow-up visit")
print("Location: Chennai")


# ---------------------------------------------------------
# 5. Feature Structure
# ---------------------------------------------------------

print("\n5. FEATURE STRUCTURE")
print("-" * 40)

subject = {
    "word": "doctor",
    "person": 3,
    "number": "singular"
}

verb = {
    "word": "recommends",
    "person": 3,
    "number": "singular"
}

print("Subject:", subject)
print("Verb:", verb)

agreement = (
    subject["person"] == verb["person"]
    and
    subject["number"] == verb["number"]
)

if agreement:
    print("Subject-Verb Agreement: Correct")
else:
    print("Subject-Verb Agreement: Incorrect")


# ---------------------------------------------------------
# 6. PCFG
# ---------------------------------------------------------

print("\n6. PCFG AMBIGUITY RESOLUTION")
print("-" * 40)

parse_probabilities = {
    "Doctor recommends two medical actions": 0.90,
    "Doctor recommends only the follow-up action": 0.10
}

for parse, probability in parse_probabilities.items():
    print(parse, "->", probability)

best_parse = max(
    parse_probabilities,
    key=parse_probabilities.get
)

print("\nMost Probable Parse:")
print(best_parse)


# ---------------------------------------------------------
# 7. Earley Parsing Concept
# ---------------------------------------------------------

print("\n7. EARLEY PARSING")
print("-" * 40)

print("Earley parsing can handle:")
print("- Long medical sentences")
print("- Ambiguous structures")
print("- Partial input")
print("- Complex grammar")

print("\nExample partial input:")
print("The doctor who reviewed the patient...")


# ---------------------------------------------------------
# 8. Sub-Categorization Frames
# ---------------------------------------------------------

print("\n8. SUB-CATEGORIZATION")
print("-" * 40)

verb_frames = {
    "recommend": ["NP", "V-ing"],
    "prescribe": ["NP"],
    "schedule": ["NP"],
    "diagnose": ["NP"],
    "monitor": ["NP"],
    "administer": ["NP"]
}

for verb, frame in verb_frames.items():
    print(verb, "->", frame)


# ---------------------------------------------------------
# 9. Extract Medical Actions
# ---------------------------------------------------------

print("\n9. MEDICAL ACTION EXTRACTION")
print("-" * 40)

actions = []

if "starting medication" in text:
    actions.append({
        "action": "start",
        "object": "medication"
    })

if "scheduling a follow-up visit" in text:
    actions.append({
        "action": "schedule",
        "object": "follow-up visit"
    })

for action in actions:
    print(action)


# ---------------------------------------------------------
# 10. Extract Time
# ---------------------------------------------------------

print("\n10. TIME EXTRACTION")
print("-" * 40)

time_match = re.search(r"last week", text)

if time_match:
    review_time = time_match.group()
else:
    review_time = None

print("Review Time:", review_time)


# ---------------------------------------------------------
# 11. Extract Location
# ---------------------------------------------------------

print("\n11. LOCATION EXTRACTION")
print("-" * 40)

location_match = re.search(r"in ([A-Z][a-z]+)", text)

if location_match:
    location = location_match.group(1)
else:
    location = None

print("Location:", location)


# ---------------------------------------------------------
# 12. Final Structured Output
# ---------------------------------------------------------

print("\n12. FINAL STRUCTURED OUTPUT")
print("-" * 40)

result = {
    "subject": "doctor",
    "diagnosis": None,
    "review_time": review_time,
    "actions": actions,
    "location": location,
    "agreement": agreement
}

print(json.dumps(result, indent=4))


# ---------------------------------------------------------
# 13. Real-Time Processing Concept
# ---------------------------------------------------------

print("\n13. REAL-TIME PROCESSING")
print("-" * 40)

print("Input received")
print("       ↓")
print("Preprocessing")
print("       ↓")
print("Medical NER")
print("       ↓")
print("CFG + Earley Parsing")
print("       ↓")
print("PCFG Ranking")
print("       ↓")
print("Semantic Analysis")
print("       ↓")
print("Structured Medical Output")


# ---------------------------------------------------------
# 14. Final Conclusion
# ---------------------------------------------------------

print("\n14. CONCLUSION")
print("-" * 40)

print("CFG provides syntactic structure.")
print("PCFG resolves ambiguity.")
print("Feature Structures check agreement.")
print("Earley Parsing handles complex sentences.")
print("Sub-categorization identifies medical relationships.")
print("Together they improve healthcare NLP accuracy and efficiency.")
```
