```python
# Semantic Representation in Customer Support Chatbots

queries = {
    "Q1": {
        "actual": ("ACTIVATE", "Roaming", "Customer"),
        "predicted": ("ACTIVATE", "Roaming", "Customer")
    },
    "Q2": {
        "actual": ("DEACTIVATE", "CallerTune", "Customer"),
        "predicted": ("ACTIVATE", "CallerTune", "Customer")
    },
    "Q3": {
        "actual": ("QUERY", "DataBalance", "Customer"),
        "predicted": ("QUERY", "DataBalance", "Customer")
    },
    "Q4": {
        "actual": ("ACTIVATE", "5GService", "Customer"),
        "predicted": ("ACTIVATE", "5GService", "Customer")
    }
}

correct = 0

for query_id, data in queries.items():

    print("\n", query_id)
    print("Actual    :", data["actual"])
    print("Predicted :", data["predicted"])

    if data["actual"] == data["predicted"]:
        print("Result    : Correct")
        correct += 1
    else:
        print("Result    : Semantic Error")

accuracy = (correct / len(queries)) * 100

print("\nTotal Correct:", correct)
print("Total Queries:", len(queries))
print("Accuracy:", accuracy, "%")
```

```python
# Smart Manufacturing using Predicate Logic

# Production data
active_machines = {"M1", "M2", "M4"}
maintenance_machines = {"M3"}

# Rule:
# Active(x) -> Producing(x)

producing_machines = set(active_machines)

# Rule:
# Maintenance(x) -> NOT Producing(x)

not_producing_machines = set(maintenance_machines)

print("Machine Status")
print("----------------")

for machine in ["M1", "M2", "M3", "M4"]:

    if machine in active_machines:
        print(machine, "-> Active -> Producing")

    elif machine in maintenance_machines:
        print(machine, "-> Maintenance -> NOT Producing")


# Product production relationships
produces = {
    "M1": ["Gear"],
    "M2": ["Shaft"],
    "M3": ["Gear"],
    "M4": ["Wheel"]
}

# Determine available products
available_products = set()

for machine, products in produces.items():

    if machine in active_machines:
        for product in products:
            available_products.add(product)

print("\nAvailable Products:")
print(available_products)


# Check whether Gear production is affected
if "M3" in maintenance_machines and "Gear" in produces["M3"]:
    print("\nGear production is affected by M3 maintenance.")
else:
    print("\nGear production is not affected by M3 maintenance.")
```

```python
# Word Sense Disambiguation

queries = {
    "Apple accessories": {
        "sense1": "Fruit",
        "sense2": "Technology Brand",
        "clicked_result": "iPhone Charger"
    },

    "Mouse wireless": {
        "sense1": "Animal",
        "sense2": "Computer Device",
        "clicked_result": "Bluetooth Mouse"
    },

    "Java tutorial": {
        "sense1": "Island",
        "sense2": "Programming Language",
        "clicked_result": "Coding Lessons"
    },

    "Python course": {
        "sense1": "Snake",
        "sense2": "Programming Language",
        "clicked_result": "Software Development Training"
    }
}


def determine_sense(query, data):

    result = data["clicked_result"].lower()

    # Technology-related clues
    technology_words = [
        "iphone",
        "charger",
        "bluetooth",
        "mouse",
        "coding",
        "software",
        "development",
        "programming",
        "lessons"
    ]

    for word in technology_words:
        if word in result:
            return data["sense2"]

    return data["sense1"]


print("Word Sense Disambiguation")
print("--------------------------")

for query, data in queries.items():

    correct_sense = determine_sense(query, data)

    print("\nQuery:", query)
    print("Clicked Result:", data["clicked_result"])
    print("Selected Sense:", correct_sense)
```

```python
# Healthcare Semantic Role Analysis

sentences = {
    "Doctor prescribed medicine to patient.": {
        "Doctor": "Agent",
        "Medicine": "Theme",
        "Patient": "Recipient"
    },

    "Patient reported severe headache.": {
        "Patient": "Experiencer/Reporter",
        "Headache": "Symptom"
    },

    "Nurse monitored patient continuously.": {
        "Nurse": "Agent",
        "Patient": "Patient/Theme"
    },

    "Medicine reduced blood pressure.": {
        "Medicine": "Cause",
        "Blood Pressure": "Affected State"
    }
}


for sentence, roles in sentences.items():

    print("\nSentence:")
    print(sentence)

    print("Semantic Roles:")

    for entity, role in roles.items():
        print(entity, "->", role)
```
